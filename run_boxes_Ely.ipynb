## Installation

Before you start, check which gcc version is installed on your machine. If it is different than gcc-11, change setup.py at lines 30 and 31. <br>

Create a separate conda environment to store this code inside it. <br>  

In this environment, you need to install the following C libraries. <br>
  ```python
    conda install -c conda-forge gsl fftw
  ```

Download 21cmFirstCLASS and checkout on the correct branch. To do so, run in the terminal the following lines.

  ```python
    git clone https://github.com/jordanflitter/21cmFirstCLASS.git 21FC
    git switch fnl
  ```

Compile the main 21cmFirstCLASS code; to do so, run in the terminal the following lines. 

  ```python
    cd 21FC/
    python setup.py clean
    python setup.py install
    pip install . 
  ```

If you do not have CLASS installed and running properly in your environment,  download it by running in the terminal the following lines.

  ```python
    git clone https://github.com/lesgourg/class_public.git CLASS
    cd CLASS/
    make
  ```
  
Open this jupyter kernel in the same env where 21cmFirstCLASS has been compiled.
